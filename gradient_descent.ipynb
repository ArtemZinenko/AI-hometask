```python
cars = {
    "Datsun PL510":     (0.97, 8.7),
    "Fiat 128":         (0.85, 8.1),
    "AMC Hornet":       (1.26, 13.1),
    "Pontiac Catalina": (2.01, 16.8),
}

def predict(w, x):
    return w * x

def error(w, x, y):
    return (predict(w, x) - y) ** 2

def slope(w, x, y):
    return 2 * (predict(w, x) - y) * x
```

```python
x, y = cars["Datsun PL510"]
w, lr = 5, 0.1

pred0 = predict(w, x)
err0 = error(w, x, y)
s = slope(w, x, y)
print(f"до кроку:    прогноз = {pred0:.4f}, помилка = {err0:.4f}, крутизна = {s:.4f}")

w_new = w - lr * s
pred1 = predict(w_new, x)
err1 = error(w_new, x, y)
print(f"w_нове = {w} - {lr} * ({s:.4f}) = {w_new:.4f}")
print(f"після кроку: прогноз = {pred1:.4f}, помилка = {err1:.4f}")
print("Стало краще?", "ТАК" if err1 < err0 else "НІ", f"(помилка {err0:.2f} -> {err1:.2f})")
```

```python
for w in (5, 9):
    print(f"--- w = {w} ---")
    for name in ("Fiat 128", "Pontiac Catalina"):
        x, y = cars[name]
        s = slope(w, x, y)
        direction = "збільшувати w" if s < 0 else "зменшувати w"
        print(f"{name:17s} прогноз = {predict(w, x):6.3f}, y = {y:5.1f}, "
              f"крутизна = {s:8.3f} -> {direction}")
```

```python
def run(lr, steps=20, w=12.0):
    path = [w]
    for _ in range(steps):
        w = w - lr * 2 * (w - 8)   # крутизна = 2*(w-8)
        path.append(w)
    return path

for lr in (0.1, 0.5, 0.9, 0.99, 1.0, 1.01, 1.1):
    p = run(lr)
    print(f"lr = {lr:<5} w: {[round(v, 3) for v in p[:6]]} ... w_20 = {p[-1]:.4g}")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
for lr in (0.1, 0.9, 1.0, 1.05):
    ax.plot(run(lr, 15), marker="o", label=f"lr = {lr}")
ax.axhline(8, color="gray", ls="--")
ax.set_xlabel("крок"); ax.set_ylabel("w"); ax.set_ylim(-5, 25)
ax.legend(); ax.set_title("Вплив lr на збіжність")
plt.show()
```

```python
x, y, w1, w2 = 1.0, 6.0, 2.0, 2.0

# 1. Прямий прохід
def forward(x, y, w1, w2):
    h = w1 * x
    pred = w2 * h
    err = (pred - y) ** 2
    return h, pred, err

h, pred, err = forward(x, y, w1, w2)
print(f"h = {h}, прогноз = {pred}, помилка = {err}")
```

```python
# 2. Зворотний прохід
d_pred = 2 * (pred - y)      # як прогноз впливає на помилку
d_w2   = d_pred * h          # крутизна по w2
d_h    = d_pred * w2         # як h впливає на помилку
d_w1   = d_h * x             # крутизна по w1

print(f"dE/dпрогноз = {d_pred}")
print(f"dE/dw2      = {d_w2}")
print(f"dE/dh       = {d_h}")
print(f"dE/dw1      = {d_w1}")
```

```python
eps = 0.001
num_w2 = (forward(x, y, w1, w2 + eps)[2] - err) / eps
num_w1 = (forward(x, y, w1 + eps, w2)[2] - err) / eps
print(f"по w2: чисельно {num_w2:.4f}  vs  аналітично {d_w2}")
print(f"по w1: чисельно {num_w1:.4f}  vs  аналітично {d_w1}")
```

```python
lr = 0.05
w1_new = w1 - lr * d_w1
w2_new = w2 - lr * d_w2
h2, pred2, err2 = forward(x, y, w1_new, w2_new)
print(f"w1 = {w1_new}, w2 = {w2_new}")
print(f"h = {h2}, прогноз = {pred2}, нова помилка = {err2:.4f}")
print("Помилка зменшилась?", "ТАК" if err2 < err else "НІ", f"({err} -> {err2:.4f})")
```
